```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d
```

```python
def plot_3d_wireframe(func, x_range, y_range, title, points=50):
    """
    func    : mathematical function f(x, y)
    x_range : tuple (xmin, xmax)
    y_range : tuple (ymin, ymax)
    title   : plot title
    points  : grid resolution
    """

    # Step 1: Define range
    x = np.linspace(x_range[0], x_range[1], points)
    y = np.linspace(y_range[0], y_range[1], points)

    # Step 2: Meshgrid
    X, Y = np.meshgrid(x, y)

    # Step 3: Function evaluation
    Z = func(X, Y)

    # Step 4: Create 3D plot
    fig = plt.figure(figsize=(8, 6))
    ax = fig.add_subplot(111, projection='3d')

    # Step 5: Wireframe plot
    ax.plot_wireframe(X, Y, Z, color='blue', linewidth=0.7)

    # Step 6: Labels
    ax.set_title(title)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_zlabel('f(x, y)')

    plt.show()
```

```python
# Q1
plot_3d_wireframe(
    lambda x, y: np.exp(x**2 + y**2),
    (0, 2*np.pi),
    (0, 2*np.pi),
    "f(x,y) = e^(x² + y²)"
)
```

```python
# Q2
plot_3d_wireframe(
    lambda x, y: np.exp(x**2 - y**2),
    (0, 2),
    (0, 2),
    "f(x,y) = e^(x² − y²)"
)
```

```python
# Q3
plot_3d_wireframe(
    lambda x, y: np.exp(-x**2 - y**2),
    (0, 2*np.pi),
    (0, 2*np.pi),
    "f(x,y) = e^(−x² − y²)"
)
```

```python
# Q4
plot_3d_wireframe(
    lambda x, y: x * np.exp(-x**2 - y**2),
    (-6, 6),
    (-6, 6),
    "f(x,y) = x e^(−x² − y²)"
)
```

```python
# Q5
plot_3d_wireframe(
    lambda x, y: np.sin(x) + np.cos(y),
    (-5, 5),
    (-5, 5),
    "f(x,y) = sin(x) + cos(y)"
)
```

```python
def plot_3d_surface(func, x_range, y_range, title, points=100):
    """
    func    : mathematical function f(x, y)
    x_range : tuple (xmin, xmax)
    y_range : tuple (ymin, ymax)
    title   : plot title
    """

    x = np.linspace(x_range[0], x_range[1], points)
    y = np.linspace(y_range[0], y_range[1], points)

    X, Y = np.meshgrid(x, y)
    Z = func(X, Y)

    fig = plt.figure(figsize=(8, 6))
    ax = fig.add_subplot(111, projection='3d')

    surf = ax.plot_surface(X, Y, Z, cmap='viridis')

    ax.set_title(title)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_zlabel('f(x, y)')

    fig.colorbar(surf, shrink=0.5, aspect=10)
    plt.show()
```

```python
plot_3d_surface(
    lambda x, y: x * np.exp(y),
    (0, 2),
    (0, 2),
    "f(x,y) = x e^y"
)
```

```python
plot_3d_surface(
    lambda x, y: x**3 + y**3 + x**2 + y**2 + x + y + 5,
    (-2, 2),
    (-2, 2),
    "Polynomial Surface"
)
```

```python
plot_3d_surface(
    lambda x, y: x**2 + y**2,
    (-6, 6),
    (-6, 6),
    "f(x,y) = x² + y²"
)
```

```python
plot_3d_surface(
    lambda x, y: np.cos(np.abs(x) + np.abs(y)),
    (-1, 1),
    (-1, 1),
    "f(x,y) = cos(|x| + |y|)"
)
```

```python
plot_3d_surface(
    lambda x, y: np.cos(x**2 + y**2 - 0.5) - 0.5,
    (-1, 1),
    (-1, 1),
    "f(x,y) = cos(x² + y² − 0.5) − 0.5"
)
```
