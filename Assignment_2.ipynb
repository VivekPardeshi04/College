```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d
```

```python
def plot_3d_contour(func, x_range, y_range, title, levels=50):
    """
    func    : function f(x, y)
    x_range : tuple (xmin, xmax)
    y_range : tuple (ymin, ymax)
    title   : plot title
    levels  : number of contour levels
    """

    # Step 1: Define ranges
    x = np.linspace(x_range[0], x_range[1], 100)
    y = np.linspace(y_range[0], y_range[1], 100)

    # Step 2: Create meshgrid
    X, Y = np.meshgrid(x, y)

    # Step 3: Compute Z
    Z = func(X, Y)

    # Step 4: Create figure and 3D axis
    fig = plt.figure(figsize=(8, 6))
    ax = plt.axes(projection='3d')

    # Step 5: Plot 3D contour
    ax.contour3D(X, Y, Z, levels, cmap='viridis')

    # Step 6: Labels and title
    ax.set_xlabel('X-axis')
    ax.set_ylabel('Y-axis')
    ax.set_zlabel('Z-axis')
    ax.set_title(title)

    # Step 7: Show plot
    plt.show()
```

```python
# Q1
plot_3d_contour(
    func=lambda x, y: x**3 + y**2,
    x_range=(0, 2*np.pi),
    y_range=(0, 2*np.pi),
    title='f(x,y) = x³ + y²'
)
```
